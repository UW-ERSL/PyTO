# FEA Non-Linear:
    TetStructuralExamples.BeamBending
    Total load = -400000
Number of nodes generated in each method varies slightly, so two columns.


```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
```

```python

ansys_nodes  = [1009, 8062, 21751, 49759, 109070]
ansys_time   = [3.3, 8.5, 18.3, 50.1, 133.7]

dpcg_nodes   = [947, 9917, 21605, 49841, 109095]
dpcg_time    = [4.955, 64.113, 164.254, 470.513, 1089.983]

pardiso_nodes = [947, 9917, 21605, 49841, 109095]
pardiso_time  = [3.508, 38.585, 91.84, 266.6566, 825.722]

plt.figure(figsize=(8,6))
plt.plot(ansys_nodes,  ansys_time,  marker='o', label='ANSYS (default)')
plt.plot(dpcg_nodes,   dpcg_time,   marker='s', label='Python DPCG')
plt.plot(pardiso_nodes, pardiso_time, marker='^', label='Python PARDISO')

plt.gca().yaxis.set_major_locator(mticker.MultipleLocator(50))
plt.xlabel('Number of nodes')
plt.ylabel('Solve time [s]')
plt.title('NonLinear analysis with Neo Hookean: Solve time vs. Number of Nodes')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
